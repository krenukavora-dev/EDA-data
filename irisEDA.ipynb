```python
fig, axes = plt.subplots(1,2,figsize=(12,5))
sns.boxplot(x='species', y = 'petal_length', data=iris, ax = axes [0])
axes [0].set_titile('Petal Length by Species')
sns.boxplot(x='species', y = 'petal_width', data=iris, ax = axes [1])
axes [1].set_titile('Petal Width by Species')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize = (12,5))
plt.subplot (1,2,1)
sns.boxplot(x='species', y = 'petal_length', data=iris)
plt.title('Petal Length Distribution by Species')
plt.ylabel('Petal Length (cm)')

plt.subplot (1,2,2)
sns.boxplot(x='species', y = 'petal_length', data=iris)
plt.title('Petal Width Distribution by Species')
plt.ylabel('Petal Width (cm)')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2,2, figsize=(12,10))
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
for i , feature in enumerate (features):
    row, col = i//2, i % 2
    sns.violinplot(data = iris, x = "species", y = feature, 
                   hue= "species", legend = False,
                   ax=axes[row, col], palette='Set2')
    axes[row, col].set_title(f'{feature.replace('_', ' ').title()} by species')
    axes[row, col].set_ylabel(feature.replace('_', ' ').title())                   
    plt.tight_layout
    plt.show()
```
